# Python 常用字符串方法

基于 `string_tools.py` 中的练习，整理常用字符串方法及其多种用法。

## 1. 大小写转换

```python
s = "Hello World"

print(s.lower())       # 全部小写: hello world
print(s.upper())       # 全部大写: HELLO WORLD
print(s.title())       # 每个单词首字母大写: Hello World
print(s.capitalize())  # 仅第一个字符大写: Hello world
print(s.swapcase())    # 大小写互换: hELLO wORLD
```

```python
# 实际应用：邮箱清洗（来自 string_tools.py）
email = " John@EXAMPLE.COM "
cleaned = email.strip().lower()
print(cleaned)  # john@example.com
```

## 2. strip() 去除空白字符

```python
s = "  hello  "

print(repr(s.strip()))   # 去除两端空白: 'hello'
print(repr(s.lstrip()))  # 只去左边: 'hello  '
print(repr(s.rstrip()))  # 只去右边: '  hello'
```

```python
# strip() 可以指定要去除的字符
s = "###hello###"
print(s.strip("#"))   # hello

s = "www.baidu.com"
print(s.strip("wmoc."))  # baidu （去除两端包含的任意字符）
```

## 3. split() 分割字符串

```python
# 默认按空白分割
s = "hello world python"
print(s.split())  # ['hello', 'world', 'python']

# 指定分隔符
url = "www.baidu.com/search?q=python"
print(url.split("/"))  # ['www.baidu.com', 'search?q=python']

# 限制分割次数
s = "a-b-c-d"
print(s.split("-", 1))  # ['a', 'b-c-d'] 只分割一次
```

```python
# 实际应用：URL域名提取（来自 string_tools.py）
url = "https://www.baidu.com/search?q=python"
domain = url.removeprefix("https://").split("/")[0]
print(domain)  # www.baidu.com
```

## 4. join() 拼接字符串

```python
words = ["hello", "world", "python"]

print(" ".join(words))   # hello world python
print("-".join(words))   # hello-world-python
print("".join(words))    # helloworldpython

# 常用于把列表拼回字符串
path_parts = ["Users", "xiangjunfeng", "Code"]
print("/".join(path_parts))  # Users/xiangjunfeng/Code
```

## 5. replace() 替换

```python
s = "hello world world"

print(s.replace("world", "python"))      # hello python python（替换所有）
print(s.replace("world", "python", 1))   # hello python world（只替换第一个）
```

```python
# 用 replace 删除字符
phone = "138-1234-5678"
print(phone.replace("-", ""))  # 13812345678
```

## 6. find() / index() 查找

```python
s = "hello world"

print(s.find("world"))    # 6 （返回起始索引）
print(s.find("python"))   # -1（找不到返回 -1）

# index() 功能相同，但找不到会报错
print(s.index("world"))   # 6
# s.index("python")       # ValueError!
```

```python
# rfind / rindex 从右边开始找
path = "/Users/xiangjunfeng/Code/test.py"
print(path.rfind("/"))  # 最后一个 / 的位置
print(path[path.rfind("/")+1:])  # test.py
```

## 7. startswith() / endswith() 判断前后缀

```python
filename = "report.pdf"

print(filename.endswith(".pdf"))    # True
print(filename.endswith(".txt"))    # False
print(filename.startswith("report")) # True

# 可以传入元组，匹配任意一个即为 True
print(filename.endswith((".pdf", ".doc", ".txt")))  # True
```

## 8. removeprefix() / removesuffix() 去除前后缀（Python 3.9+）

```python
url = "https://www.baidu.com"
print(url.removeprefix("https://"))  # www.baidu.com
print(url.removeprefix("http://"))   # 不匹配则原样返回

filename = "test.py"
print(filename.removesuffix(".py"))  # test
```

## 9. 判断类方法：isdigit() / isalpha() / isalnum()

```python
print("123".isdigit())    # True 全是数字
print("12.3".isdigit())   # False 小数点不算
print("abc".isalpha())    # True 全是字母
print("abc123".isalnum()) # True 字母+数字
print("  ".isspace())     # True 全是空白字符
```

```python
# 实际应用：简单输入验证
user_input = "abc123"
if user_input.isalnum():
    print("有效的用户名")
else:
    print("用户名只能包含字母和数字")
```

## 10. count() / len() 计数

```python
s = "hello world hello python"

print(s.count("hello"))  # 2 出现次数
print(len(s))            # 24 字符串长度（len是内置函数，不是方法）
```

## 11. zfill() / center() / ljust() / rjust() 填充对齐

```python
# zfill 用0填充到指定宽度
print("42".zfill(5))     # 00042

# center / ljust / rjust
s = "hi"
print(s.center(10, "-"))  # ----hi----
print(s.ljust(10, "-"))   # hi--------
print(s.rjust(10, "-"))   # --------hi
```

## 12. 链式调用

字符串方法返回的还是字符串，所以可以连续调用多个方法：

```python
raw = "  HTTPS://WWW.EXAMPLE.COM/Path  "

# 链式调用：去空白 → 转小写 → 去前缀 → 取域名
domain = raw.strip().lower().removeprefix("https://").split("/")[0]
print(domain)  # www.example.com

# 注意：split() 返回列表，链就断了，不能再接字符串方法
# 但可以接列表方法或用索引取值后继续
```
